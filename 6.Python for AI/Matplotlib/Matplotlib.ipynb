 ***Matplotlib***

```python
import numpy as np
import matplotlib.pyplot as plt
```

***Line and Scatter Plots***

```python
# 1. Generate data
x = np.linspace(0, 10, 50)  # 50 evenly spaced points between 0 and 10
y1 = np.sin(x)
y2 = np.cos(x)

# 2. Create the canvas
fig, ax = plt.subplots(figsize=(8, 5))

# 3. Plot the data
ax.plot(x, y1, label='Sine Wave', color='blue', linestyle='--')
ax.scatter(x, y2, label='Cosine Dots', color='red', marker='x')

# 4. Add labels (The Museum Placard)
ax.set_title("Wave Functions Over Time")
ax.set_xlabel("Time (seconds)")
ax.set_ylabel("Amplitude")
ax.legend()
ax.grid(True)

# 5. Show the plot
plt.show()
```

***The Bar Chart***

```python
# 1. Generate categorical data
categories = ['Apples', 'Bananas', 'Cherries', 'Dates']
sales = [25, 40, 15, 30]

# 2. Create the canvas
fig, ax = plt.subplots(figsize=(8, 6))

# 3. Plot the bar chart
ax.bar(categories, sales, color='skyblue', edgecolor='black')

# 4. Add labels
ax.set_title("Store Sales Comparison")
ax.set_xlabel("Fruit Type")
ax.set_ylabel("Units Sold")

# 5. Show the plot
plt.show()
```

***The Histogram***

```python
# 1. Generate 1000 random data points using a normal (Gaussian) distribution
random_distribution = np.random.randn(1000)

# 2. Create the canvas
fig, ax = plt.subplots(figsize=(6, 4))

# 3. Plot the histogram
# 'bins' determines how many bars the data is split into
ax.hist(random_distribution, bins=20, color='green', alpha=0.7)

# 4. Add labels
ax.set_title("Data Frequency Distribution")
ax.set_xlabel("Value")
ax.set_ylabel("Frequency (Count)")

# 5. Show the plot
plt.show()
```

***Advanced : Subplots***

```python
# 1. Generate data
x = np.arange(1, 6)
y_squared = x ** 2
y_cubed = x ** 3

# 2. Create a canvas with 1 row and 2 columns
# ax1 is the left plot, ax2 is the right plot
fig, (ax1, ax2) = plt.subplots(nrows=1, ncols=2, figsize=(10, 4))

# 3. Paint the Left Plot (ax1)
ax1.plot(x, y_squared, color='green', marker='o')
ax1.set_title("X Squared")
ax1.set_xlabel("X")
ax1.set_ylabel("Y")

# 4. Paint the Right Plot (ax2)
ax2.plot(x, y_cubed, color='orange', marker='s')
ax2.set_title("X Cubed")
ax2.set_xlabel("X")
ax2.set_ylabel("Y")

# 5. Clean up the layout and show
plt.tight_layout() # This prevents the labels from overlapping
plt.show()
```
