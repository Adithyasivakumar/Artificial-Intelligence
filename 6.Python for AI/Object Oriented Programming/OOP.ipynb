***Inheritance***

Inheritance



*   Allows a class to acquire attributes(properties) and methods form another class
*   Helps with code reusablity and extensibility
*   Child class is inherited from parent class



```python
class Animal: # parent class

  def __init__(self, name): # constructor method

    self.name = name
    self.is_alive = True

  def eat(self): # method
    print(f"The {self.name} is eating")

  def sleep(self): # method
    print(f"The {self.name} is sleeping")

class Dog(Animal): # child class 1
  pass

class Cat(Animal): # child class 2
  pass

class Mouse(Animal): # child class 3
  pass

#objects

dog = Dog("Scooby")
cat = Cat("Mieu")
mouse = Mouse("Hitler")

print(dog.name)
print(cat.name)
print(dog.is_alive)
dog.eat()
mouse.sleep()


# If you dont use inheritance, You have to write the same amount of code written
# for main class to all the other classes. If we make any changes in one class
# we have to simultaneously make changes in the other class too. Which is very
# complicated


# Core idea : Write code once and reuse it!
```

***Multiple Inheritance | Multilevel Inheritance***

***Multiple Inheritance*** : Inherits from more than one parent class -> C(A,B)

***Multilevel Inheritance*** : Inherits from a parent which inherits from another parent -> C(B) <- B(A) <- A


```python
# Multiple Inheritance

class Prey: # parent class 1

  def flee(self):
    print("The animal is fleeing")

class Predator: # parent class 2

  def hunt(self):
    print("The animal is hunting")

class Rabbit(Prey): # child class 1
  pass

class Hawk(Predator): # child class 2
  pass

class Fish(Prey, Predator): # child class 3 (Multiple Inheritance)
  pass

# creating objects

rabbit = Rabbit()
hawk = Hawk()
fish = Fish()

rabbit.flee() # it can only flee
hawk.hunt() # it can only hunt
fish.flee() # it can flee
fish.hunt() # as well as hunt
```

```python
# multilevel Inheritance

class Animal: # grand parent

  def eat(self):
    print("The animal is eating")

  def sleep(self):
    print("The animal is sleeping")

class Prey(Animal): # parent class 1 - Inherits Animal class

  def flee(self):
    print("The animal is fleeing")

class Predator(Animal): # parent class 2 - Inherits Animal class

  def hunt(self):
    print("The animal is hunting")

class Rabbit(Prey): # child class 1 - Inherits Pery class
  pass

class Hawk(Predator): # child class 2 - Inherits Predator class
  pass

class Fish(Prey, Predator): # child class 3 (Multiple Inheritance) - Inherits both Prey and Predator class
  pass

# creating objects

rabbit = Rabbit()
hawk = Hawk()
fish = Fish()

rabbit.eat()
rabbit.flee()
fish.sleep()
fish.hunt()
fish.flee()
```

```python
# multilevel Inheritance

class Animal: # grand parent

  def __init__(self, name): # constructor method
    self.name = name

  def eat(self):
    print(f"{self.name} is eating")

  def sleep(self):
    print(f"{self.name} is sleeping")

class Prey(Animal): # parent class 1 - Inherits Animal class

  def flee(self):
    print(f"{self.name} is fleeing")

class Predator(Animal): # parent class 2 - Inherits Animal class

  def hunt(self):
    print(f"{self.name} is hunting")

class Rabbit(Prey): # child class 1 - Inherits Pery class
  pass

class Hawk(Predator): # child class 2 - Inherits Predator class
  pass

class Fish(Prey, Predator): # child class 3 (Multiple Inheritance) - Inherits both Prey and Predator class
  pass

# creating objects

rabbit = Rabbit("Shushi")
hawk = Hawk("Oddy")
fish = Fish("Nemo")

print(rabbit.name)
print(hawk.name)
print(fish.name)
rabbit.eat()
fish.sleep()
hawk.hunt()
```

***ABSTRACT CLASSES***


*   A class that cannot be instantiated on it's own; Meant to be subclassed
*   They can contain abstract methods, which are declared but have no implementation

***Benefits***

*   Prevents instantiation of the class itself
*   Requires children to use inherited abstract methods

```python
# To deal with the abstract methods, we need to import the Abstract Base Class (ABC)

from abc import ABC, abstractmethod

class Vehicle(ABC):

  @abstractmethod #decorator
  def go(self):
    pass

  @abstractmethod #decorator
  def stop(self):
    pass

class Car(Vehicle):

  def go(self):
    print("You drive the car") # method 1

  def stop(self):
    print("You stop the car") # method 2

class Bike(Vehicle):

  def go(self):
    print("You ride the bike") # method 1

  def stop(self):
    print("You stop the bike") # method 2

class Boat(Vehicle):

  def go(self):
    print("You start the boat") # method 1

  def stop(self):
    print("You stop the boat") # method 2 (If we remove this method and run the code it will throw an error because the abstract class have two abstract methods, and you have initialized only one method)

car = Car() # object
bike = Bike() # object
boat = Boat() # object

car.go()
car.stop()

bike.go()
bike.stop()

boat.go()
boat.stop()
```

***Super()***


*   It is a function used in child class (subclass) to call methods from a parent class (superclass)
*   Allows you to extend the functionality of the inherited methods



```python
class Circle:

  def __init__(self, color, is_filled, radius):

    self.color = color # repeated
    self.is_filled = is_filled # repeated
    self.radius = radius

class Square:

  def __init__(self, color, is_filled, width):

    self.color = color # repeated
    self.is_filled = is_filled # repeated
    self.width = width

class Triangle:

  def __init__(self, color, is_filled, width, height):

    self.color = color # repeated
    self.is_filled = is_filled # repeated
    self.width = width
    self.height = height

# since same attributes are repeated again and again we can declare it in the main class and use the super() function to access those attributes
```

```python
class Shape: # parent class

  def __init__(self, color, is_filled):

    self.color = color
    self.is_filled = is_filled

class Circle(Shape):

  def __init__(self, color, is_filled, radius):

    super().__init__(color, is_filled) # here we have called the constructor method form the parent class with the help of super() func
    self.radius = radius

class Square(Shape):

  def __init__(self, color, is_filled, width):

    super().__init__(color, is_filled)
    self.width = width

class Triangle(Shape):

  def __init__(self, color, is_filled, width, height):

    super().__init__(color, is_filled)
    self.width = width
    self.height = height

#objects

circle = Circle("Black", True, 5)
square = Square("Blue", False, 7)
triangle = Triangle("Grey", True, 8, 9)

print(circle.color)
print(circle.is_filled)
print(f"{circle.radius}cm")

print(triangle.color)
print(triangle.is_filled)
print(f"{triangle.width}cm, {triangle.height}cm")
```

```python
# Extending the access
# We will create a new method inside the parent class, and when we use the super() the subclass have access to all the methods including newly created ones

class Shape: # parent class

  def __init__(self, color, is_filled):

    self.color = color
    self.is_filled = is_filled

  def describe(self):
    print(f"It is {self.color} and {'filled' if self.is_filled else 'not filled'}")

class Circle(Shape):

  def __init__(self, color, is_filled, radius):

    super().__init__(color, is_filled) # here we have called the constructor method form the parent class with the help of super() func
    self.radius = radius

class Square(Shape):

  def __init__(self, color, is_filled, width):

    super().__init__(color, is_filled)
    self.width = width

class Triangle(Shape):

  def __init__(self, color, is_filled, width, height):

    super().__init__(color, is_filled)
    self.width = width
    self.height = height

#objects

circle = Circle("Black", True, 5)
square = Square("Blue", False, 7)
triangle = Triangle("Grey", True, 8, 9)

print(circle.color)
print(circle.is_filled)
print(f"{circle.radius}cm")

circle.describe()
square.describe()
triangle.describe()
```

***Method Overriding***

```python
# Same above example

class Shape: # parent class

  def __init__(self, color, is_filled):

    self.color = color
    self.is_filled = is_filled

  def describe(self): # describe 1
    print(f"It is {self.color} and {'filled' if self.is_filled else 'not filled'}")

class Circle(Shape):

  def __init__(self, color, is_filled, radius):

    super().__init__(color, is_filled) # here we have called the constructor method form the parent class with the help of super() func
    self.radius = radius

  def describe(self): # describe 2
    print(f"It is a circle with an area of {3.14 * self.radius * self.radius}cm^2")
    super().describe() # I can also access the describe function which is defined inside the main class using the super() func

class Square(Shape):

  def __init__(self, color, is_filled, width):

    super().__init__(color, is_filled)
    self.width = width

  def describe(self): # describe 2
    print(f"It is a square with an area of {self.width * self.width}cm^2")
    super().describe() # I can also access the describe function which is defined inside the main class using the super() func

class Triangle(Shape):

  def __init__(self, color, is_filled, width, height):

    super().__init__(color, is_filled)
    self.width = width
    self.height = height

  def describe(self): # describe 2
    print(f"It is a triangle with an area of {0.5 * self.width * self.height}cm^2")
    super().describe() # I can also access the describe function which is defined inside the main class using the super() func

#objects

circle = Circle("Black", True, 5)
square = Square("Blue", False, 7)
triangle = Triangle("Grey", True, 7, 8)

circle.describe() # There are two describe methods declared and the method inside the subclass is executed
square.describe()
triangle.describe()
```

***Polymorphism***

*   Greek word that means "to have many forms or faces"
*   Poly = many | morphe = forms

***Ways to achieve***

*   Inheritance : An object could be treated of the same type as a parent class
*   Duck Typing : Object must have necessary attributes / methods





```python
# Polymorphism using Inheritance

class Shape:
  pass

class Circle(Shape):
  pass

class Square(Shape):
  pass

class Triangle(Shape):
  pass

circle = Circle()

# In this case the circle have two forms : Circle and Shape. But not Square and Triangle
```

```python
# Polymorphism using Inheritance

from abc import ABC, abstractmethod

class Shape: # parent class (Abstract class)

  @abstractmethod
  def area(self):
    pass

class Circle(Shape): # child class 1

  def __init__(self, radius): # constructor method
    self.radius = radius

  def area(self):
    return 3.14 * self.radius ** 2

class Square(Shape): # child class 2

  def __init__(self, side): # constructor method
    self.side = side

  def area(self):
    return self.side * self.side

class Triangle(Shape): # child class 3

  def __init__(self, base, height): # constructor method
    self.base = base
    self.height = height

  def area(self):
    return 0.5 * self.base * self.height

circle = Circle(4)
square = Square(5)
triangle = Triangle(6,7)

print(f"{circle.area()}cm²")
print(f"{square.area()}cm²")
print(f"{triangle.area()}cm²")
```

```python
# Polymorphism using Inheritance

from abc import ABC, abstractmethod

class Shape: # parent class (Abstract class)

  @abstractmethod
  def area(self):
    pass

class Circle(Shape): # child class 1

  def __init__(self, radius): # constructor method
    self.radius = radius

  def area(self):
    return 3.14 * self.radius ** 2

class Square(Shape): # child class 2

  def __init__(self, side): # constructor method
    self.side = side

  def area(self):
    return self.side * self.side

class Triangle(Shape): # child class 3

  def __init__(self, base, height): # constructor method
    self.base = base
    self.height = height

  def area(self):
    return 0.5 * self.base * self.height

class Pizza(Circle): # child class 4 : Inherits from the Circle class since pizza is circle

  def __init__(self, toppings, radius): # constructor
    super().__init__(radius) # super function
    self.toppings = toppings

shapes = [Circle(4), Square(5), Triangle(6,7), Pizza('Pepperoni', 15)]

for shape in shapes:

  print(f"{shape.area()}cm²")


# In case of Pizza : It is considered as a Pizza, as well as a Circle, and it is also considered as a Shape
# It has multiple forms : Pizza, Circle, Shape
```

***Duck Typing***

*   Another way to achieve polymorphism beside Inheritance

*   Object must have minimum necessary attributes/methods

*   "If it look like a duck and quacks like a duck, it must be a duck"


```python
class Animal:
  alive = True

class Dog(Animal):

  def speak(self):
    print('WOOF!')

class Cat(Animal):

  def speak(self):
    print("MEOW!")

animals = [Dog(), Cat()]

for animal in animals:
  animal.speak()
```

***Aggregation***

*  Represents a relationship where one object (the whole) contains references to one or more INDEPENDENT objects (the parts)


```python
# class 1

class Library:
  def __init__(self, name):
    self.name = name
    self.books = []

  # adding books
  def addBook(self, book):
    self.books.append(book)

  # print books inside the library
  def list_books(self):
    return [f"{book.title} by {book.author}" for book in self.books]

# class 2

class Book:
  def __init__(self, title, author):
    self.title = title
    self.author = author

# Creating objects

library = Library("The central library")

book1 = Book("Harry Potter", "J.K. Rowling")
book2 = Book("1984", "George Orwell")
book3 = Book("The Alchemist ", "Paulo Coelho")


library.addBook(book1)
library.addBook(book2)
library.addBook(book3)

print(library.name)

# print(library.list_books())  --straight forward way of displaying elements

for book in library.list_books():
  print(book)


```

***Composition***

*  Aggregation : A relationship where one object contains references to other independent objects "has-a" relationship
*  Composition : The composed object directly owns its components, which cannot exist independently "owns-a" relationship



```python
class Engine:

  def __init__(self, horse_power):
    self.horse_power = horse_power

class Wheel:

  def __init__(self, size):
    self.size = size

class Car:

  def __init__(self, make, model, horse_power, wheel_size):

    self.make = make
    self.model = model
    self.engine = Engine(horse_power) # Renamed to avoid conflict
    self.wheels = [Wheel(wheel_size) for _ in range(4)] # Use _ for unused loop variable

  def display_car(self):

    return f"Make: {self.make}, Model: {self.model}, Horse Power: {self.engine.horse_power}, Wheel Size: {self.wheels[0].size}"

# creating objects

car = Car("Honda", "Amaze", 100, 17)

print(car.display_car())
```

***Nested Class***

*   A class defined within another class

Benefits:

*   Allows you to logically group classes that are closely related
*   Encapsulates private details that aren't relevent outside of the outer class
*   Keeps the namespace clean; reduces the possiblity of **naming conflicts**



```python
class Company:

  #constructor
  class Employee:

    def __init__(self, name, position):

      self.name = name
      self.position = position

    # method to display details
    def get_details(self):
      return f"{self.name} {self.position}"

  #constructor
  def __init__(self, company_name):
    self.company_name = company_name
    self.employees = []

  # method for adding employees
  def add_employees(self, employee_name, position):

    new_employee = self.Employee(employee_name, position)
    self.employees.append(new_employee)

  # method for displaying the employees
  def list_employees(self):
    return [employee.get_details() for employee in self.employees]

company1 = Company("Deloitte")
company2 = Company("EY")

company1.add_employees("Adithya", "Manager")
company1.add_employees("Harini", "Tester")
company1.add_employees("Nhowmi", "HR")

company2.add_employees("Peter", "Manager")
company2.add_employees("Alice", "Tester")
company2.add_employees("Bob", "HR")

company.list_employees()

for employee in company2.list_employees():
  print(employee)
```
