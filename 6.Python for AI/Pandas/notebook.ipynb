# 4. Displaying & Filtering

```python
df.head()
```

```python
# Filtering all the Electronics

df[df["Category"] == "Electronics"]
```

```python
# Filtering based on multiple values using logical and (&)

# similarly we can perform all arithmentic and logical operations

df[(df["Category"] == "Electronics") & (df["Price"] >= 100)]
```

# Advance Filtering

```python
# 1. Displays all the customer names which starts with "A"

df[df["CustomerName"].str.startswith('A')]
```

```python
# 2. Similarly, Displays all the customer names which ends with "e"

df[df["CustomerName"].str.endswith("e")]
```

```python
# To display datas based on conditions

df[df["Country"].isin(["USA", "France", "India"])]
```

```python
# To display specific information, based on not in condition (~)

df[~df["Country"].isin(["USA", "France", "India"])]
```

# Updating data

```python
# Locating the data

df.loc[df["CustomerName"] == "Anna Ivanova"]
```

```python
# changing the value of the Product

df.loc[df["CustomerName"] == "Anna Ivanova", "Product"] = "Pen"
```

```python
# Result : Value is changed

df.loc[df["CustomerName"] == "Anna Ivanova"]
```

```python
# Another example : Changing the values of the Country

df.loc[df["Country"] == "USA", "Country"] = "United States of America"
```

```python
# Result : Value is changed

df.loc[df["Country"] == "United States of America"]
```

```python
# Converting values into Uppercase : Similarly we can do this to convert into lowercase and soon

df["Country"] = df["Country"].str.upper()
```

```python
df["Country"]
```

# Deleting data

```python
# To delete specific rows

df = df.drop(39)
```

```python

df.tail()
```

# Cleaning data

```python
# Drop anything that contains any null values or na values

df.dropna()
```

```python
# To fill in entries with null values

df.fillna({"OrderID": 0}, inplace = True) # modifies inplace, dosen't return anything
```

```python
# Renaming specific column

df.rename(columns={"OrderID": "Order ID"}, inplace = True)
```

```python
df.head()
```

# Analyzing data

```python
# 1. value_counts : provide the individual count of each value

df["Country"].value_counts()
```

```python
# 2. groupby : This gives the total sum for each country

df.groupby("Country")["Price"].sum()
```

```python
# 3. sort_values : This sort the values in ascending or descending order

df.sort_values("Price", ascending = False)
```
